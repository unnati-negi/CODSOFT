```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
!pip install hmmlearn
from hmmlearn import hmm
```

```python
data = pd.read_csv("/content/Heart_Failure_Details.csv")

data.head()
```

```python
data.describe()
```

```python
marginal_prob = data['death'].mean()

print("P(Death Event) =", marginal_prob)
```

```python
joint_prob = len(data[(data['age'] > 60) & (data['death'] == 1)]) / len(data)

print("P(Age > 60 AND Death = 1) =", joint_prob)
```

```python
age = data[data['age'] > 60]

conditional_prob = age['death'].mean()

print("P(Death | Age > 60) =", conditional_prob)

if conditional_prob > marginal_prob:
    print("Older patients have higher mortality risk.")
else:
    print("Older patients do not have higher mortality risk.")
```

```python
mle = data['death'].sum() / len(data)

print("MLE Estimate of Death Probability =", mle)
```

```python
p = mle
q = 0.25

kldiv = p*np.log(p/q) + (1-p)*np.log((1-p)/(1-q))

print("KL Divergence =", kldiv)
```

```python
states = ["Healthy", "At Risk", "Critical"]

transition_matrix = np.array([
    [0.7, 0.2, 0.1],
    [0.3, 0.5, 0.2],
    [0.1, 0.3, 0.6]
])

current_state = 0

print("Patient Health Progression")

for day in range(10):
    current_state = np.random.choice([0,1,2], p=transition_matrix[current_state])
    print("Day", day, "State:", states[current_state])
```

```python
health_sequence = data['death'].values.reshape(-1,1)

model = hmm.MultinomialHMM(n_components=2)

model.fit(health_sequence)

hidden_states = model.predict(health_sequence)

print("First 20 Hidden States:", hidden_states[:20])
```
